```python
# Check Ollama server availability
import requests
base_url = 'http://localhost:11434'
try:
    r = requests.get(f'{base_url}/api/tags', timeout=5)
    r.raise_for_status()
    print('Ollama server is up:', r.json())
except Exception as e:
    print('Ollama server not reachable:', e)
```

```python
# Simple prompt to gemma model
import requests, json
base_url = 'http://localhost:11434'
payload = {
    'model': 'gemma4:e2b',
    'prompt': 'Translate to English: こんにちは',
    'temperature': 0.0,
    'stream': False,
}
try:
    r = requests.post(f'{base_url}/api/generate', json=payload, timeout=120)
    r.raise_for_status()
    resp = r.json()
    print('Raw response:', json.dumps(resp, ensure_ascii=False, indent=2))
    print('\nExtracted text:\n', resp.get('response', resp))
except Exception as e:
    print('Request failed:', e)
```
